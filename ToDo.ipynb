```python
#To-Do list allowing users to: 
# 1. Add a list 
# 2. View List
# 3. Delete List 
# 4. Mark as Finish 
# 5. Choose Options 
# 6. Delete entire list
```

```python
tasks = []
```

```python
# Helper Function 
def UI(): 
    print("|----- TO-DO -----|")
    print("\n1. Add Task")
    print("2. View Tasks")
    print("3. Mark As Done")
    print("4. Delete Task")
    return

def add_task(): 
    task = input("Input Task: ") 
    tasks.append({"Task": task, "Status": False})
    print(f"{task} Added!\n") 
    return

def view_task():
    if not tasks: 
        print("No Tasks Yet!\n") 
        UI()
        return
        
    for i, task in enumerate(tasks, start=1): 
        status = "Done" if task["Status"] == True else "Not Done"
        print(f"{i}. {task["Task"]}   [ {status} ]")
    return

def delete_task(): 
    if not tasks: 
        print("No Tasks Yet!\n") 
        UI()
        return
        
    view_task()
    task_index = int(input("Delete Task Number: "))
    tasks.pop(task_index - 1)
    return 

def mark_finish(): 
    view_task() 
    task_index = int(input("Task Finished: ")) - 1
    tasks[task_index]["Status"] = True

    return
```

```python
# Main Loop 

while True: 
    UI()
    choice = int(input("\nChoose An Option (number): "))

    if choice == 1: 
        add_task()

    elif choice == 2: 
        view_task()

    elif choice == 3: 
        mark_finish()

    elif choice == 4: 
        delete_task()

    else:
        print("\nInvalid Input. Enter Option Number.\n")
```
