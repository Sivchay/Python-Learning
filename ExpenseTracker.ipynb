```python
# Project is to learn how to store data using JSON

# Add_Expense: asks for {description/type, amount}  - eg. {"Lunch", 12} 
# View_Expense: iterates over list containing dict with expenses 
# Delete_Expense: Uses View_Expense to show all expenses and deletes the index of dict 
# View_Total: Calculates cost of by summing [i]["amount"] 
```

```python
expenses  = [] 
```

```python
def add_expense(): 
    name  = input("Description of Item? (Lunch, Gas, etc.)")
    amount  = int(input("Item Cost?")) 

    expense  = {
        
    "Type": name, 
    "Cost": amount
        
    }
    expenses.append(expense)
    
```

```python
add_expense()
```
