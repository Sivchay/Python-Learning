```python
import random


deck        = {
    "A":  11,
    "K":  10,
    "Q":  10,
    "J":  10,
    "10": 10,
    "9":   9,
    "8":   8,
    "7":   7, 
    "6":   6,
    "5":   5,
    "4":   4,
    "3":   3,
    "2":   2, 
    "?":   0,
}
```

```python
#Helper Functions 
def UI(): 
    print("\n\n|---------- Black Jack ----------|") 
    print("\nDealer's Hand") 
    print(f"{dealer_hand}    Value: {dealer_calc()}")
    print("\nYour Hand")
    print(f"{player_hand}    Value: {player_calc()}")

def create_hand(): 
    dealer_hand = None
    player_hand = None
    dealer_hand = [random.choice(list(deck.keys())), "?"]
    player_hand = random.choices(list(deck.keys()), k=2)
    
    return dealer_hand, player_hand

def player_choice(): 
    #Hit = A
    #Stand = B

    while True:
        UI()
        choice = str(input("\nHit [A] or Stand [B]?")).lower()
        print("\n|--------------------------------|") 
    
        if choice == "a": 
            player_hand.append(random.choice(list(deck.keys())))
    
        elif choice == "b": 
            dealer_choice()

        else: 
            print("Enter Valid Choice!\n") 
            
def dealer_choice(): 
    dealer_hand.pop()
    dealer_hand.append(random.choice(list(deck.keys())))
    dealer_count = dealer_calc() 
    
    while dealer_count >= 17 or dealer_count <= 21: 
        dealer_hand.append(random.choice(list(deck.keys())))
        dealer_count = dealer_calc() 

    else: 
        dealer_count = dealer_calc() 
    
        
def dealer_calc(): 
    dealer_count = sum(deck[key] for key in dealer_hand)
    aces         = True if "A" in dealer_hand else False

    if dealer_count > 21 and aces: 
        dealer_count -= 10 
        
    return dealer_count

def player_calc(): 
    player_count = sum(deck[key] for key in player_hand)
    aces         = True if "A" in player_hand else False

    if player_count > 21 and aces: 
        player_count -= 10 

    return player_count
        
        
```
