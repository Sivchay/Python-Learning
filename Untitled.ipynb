```python
catalouge = [] 
```

```python
#Helper Functions 

def UI(): 
    print("|----- Library Catalouge -----|")
    print("\n1. Add Books")
    print("2. Borrow Books")
    print("3. Return Books")
    print("4. View Books")
    print("5. Search Books")
    
    
def add_book(): 
    book_title = input("Book Title: ") 
    author     = input("Author: ") 
    catalouge.append({"Book Title": book_title, "Author": author, "Status": True}) 
    print(f"{book_title} added!")

def view_books(): 
    for i, book in enumerate(catalouge, start=1): 
        status = "Availiable" if book["Status"] else "Borrowed"
        print(f"\n{i}. {book["Book Title"]} \nAuthor: {book["Author"]} \nStatus: [ {status} ]") 

def borrow_book(): 
    view_books() 
    index      = int(input("Select Book Number: ")) - 1 
    catalouge[index]["Status"] = False if catalouge[index]["Status"] else print("Already Borrowed")
    

def return_book(): 
    view_books()
    index      = int(input("Select Book Number: ")) - 1 
    catalouge[index]["Status"] = print("Already Returned") if catalouge[index]["Status"] else True
    

def search_books(): 
    search     = input("Search: ").lower()
    result     = False
    
    for i, book in enumerate(catalouge, start=1): 
        if book["Book Title"].lower() == search: 
            status = "Availiable" if book["Status"] else "Borrowed"
            print(f"\n{i}. {book["Book Title"]} \nAuthor: {book["Author"]} \nStatus: [ {status} ]") 
            result = True

        elif result is False: 
            print("No Results")
            break

    


    
```

```python
while True: 
    UI()
    choice = int(input("\nChoose An Option (number): "))

    if choice == 1: 
        add_book()

    elif choice == 2: 
        borrow_book()

    elif choice == 3: 
        return_book()

    elif choice == 4: 
        view_books()

    elif choice == 5: 
        search_books()
    else:
        print("\nInvalid Input. Enter Option Number.\n")
```
