```python
from datetime import time, datetime
```

```python
def hint(): 
    if guess > number:
        return print("(Hint) Slightly Lower") if (guess - number) <= 2 else print("(Hint) Lower")

    elif guess < number: 
        return print("(Hint) Slightly Higher") if (number - guess) <= 2 else print("(Hint) Higher")
```

```python
now    = str(datetime.now())
number = int(now[-2])
lives  = 3
guess  = int(input(f"What is the number? \nLives Left: {lives} \nYour Guess: "))

while guess != number: 
    if lives >= 1: 
        hint()
        lives -= 1
        guess  = int(input(f"\nTry again \nLives Left: {lives} \nYour Guess: "))

    else: 
        print("\nOut of Lives :(") 
        break 

else: 
    print("Correct!") 

```
